```python
import numpy as np
import json
import matplotlib.pyplot as plt
import pandas
# import chainconsumer
# from chainconsumer import ChainConsumer, Chain, PlotConfig, Truth
import corner
plt.rcParams.update({'font.size': 15})
import matplotlib
matplotlib.rcParams['text.usetex'] = True

%config InlineBackend.figure_format = 'retina'
plt.rc('text', usetex=True)
plt.rc('font', family='serif')
```

```python
with open('/home/ansonchen/multiband_cosmo/code/out/boson_null_a_2/boson_null_a_2_result.json','r')as file:
    data = json.load(file)
```

```python
data['posterior']['content'].keys()
```

```python
label = [r'$({\cal M}_c/M_\odot-30)\,[\times 10^{-4}]$', r'$q$', r'$t_c/s$', r'$\varepsilon_c^0\,[\times 10^{-4}]$']

val = np.array([(np.array(data['posterior']['content']['Mc'])-30)*1e4, data['posterior']['content']['q'], np.array(data['posterior']['content']['t_c_sec'])-4*365.25*24*3600, np.array(data['posterior']['content']['Mc0'])*1e4]).T

truth = [0.0, 0.9, 0.0, 0.0]
```

```python
plt.figure()
fig = corner.corner(val, labels=label, truths=truth, truth_color='darkorange', bins=35, smooth=1, label_kwargs=dict(fontsize=16), color='#0072C1', levels=(1-np.exp(-0.5),1-np.exp(-2),1-np.exp(-4.5)), quantiles=[0.16, 0.5, 0.84], show_titles=True, title_kwargs={"fontsize": 15})
plt.savefig('out/boson_null_a_2_corner.png',dpi=600)
```
